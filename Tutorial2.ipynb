```python
#1. Create a numpy array of size of 2 x 2.


import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size of 3 x 4.


random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes.



ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Arrays of Ones:")
print(ones_array)

print("Array of Zeros:")
print(zeros_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it.


random6x3 = np.random.rand(6,3)
transposed_array = np.transpose(random6x3)

print("Random 6x3 array:")
print(random6x3)

print("Transposed array:")
print(transposed_array)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product.


n, p, m = 3, 4, 2

matrix_A = np.random.rand(n,p)
matrix_B = np.random.rand(p,m)

dot_product = np.dot(matrix_A, matrix_B)

print("Matrix A:")
print(matrix_A)

print("Matrix B:")
print(matrix_B)

print("Dot Product:")
print(dot_product)
```
